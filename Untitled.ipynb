```python
import pandas as pd
import numpy as np
```

```python
trans_mat1 = pd.read_csv('hmm_trans_matdf.csv')
trans_mat1.index = trans_mat1['Unnamed: 0']
trans_mat1.drop('Unnamed: 0', axis=1, inplace=True)

trans_mat2 = pd.read_csv('hqmm_trans_matdf.csv')
trans_mat2.index = trans_mat2['Unnamed: 0']
trans_mat2.drop('Unnamed: 0', axis=1, inplace=True)
```

```python
np.sum(trans_mat1.to_numpy() - trans_mat2.to_numpy())
```
