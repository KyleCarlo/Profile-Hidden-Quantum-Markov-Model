```python
import pyhmmer

alphabet = pyhmmer.easel.Alphabet.amino()

# Load profile HMM
with pyhmmer.plan7.HMMFile("model.hmm") as hmm_file:
    hmm = next(hmm_file)

# Load protein sequences
with pyhmmer.easel.SequenceFile(
    "sequences.fasta",
    digital=True,
    alphabet=alphabet
) as seq_file:
    sequences = seq_file.read_block()

# Align sequences against the HMM
msa = pyhmmer.hmmer.hmmalign(hmm, sequences)

# Save alignment in Stockholm format
with open("aligned.sto", "wb") as f:
    msa.write(f, format="stockholm")
```
