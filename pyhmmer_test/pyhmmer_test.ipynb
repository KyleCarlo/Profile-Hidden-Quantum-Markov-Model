```python
import pyhmmer

alphabet = pyhmmer.easel.Alphabet.amino()

# Load profile HMM
with pyhmmer.plan7.HMMFile("PF30808.hmm") as hmm_file:
    hmm = next(hmm_file)

# Load protein sequences
with pyhmmer.easel.SequenceFile(
    "A0A010YIN8.fa",
    digital=True,
    alphabet=alphabet
) as seq_file:
    sequences = seq_file.read_block()

# Score sequences against the profile HMM
results = pyhmmer.hmmer.hmmsearch(
    [hmm],
    sequences,
    T=-1000.0,
    incT=-1000.0
)

for hits in results:
    for hit in hits:
        print(f"Sequence: {hit.name}")
        print(f"Bit score: {hit.score}")
        print(f"E-value: {hit.evalue}")
```
